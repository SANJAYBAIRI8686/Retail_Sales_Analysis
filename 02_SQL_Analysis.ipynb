## Total Sales 

### Business Question: How much money did we make in total?

```python
import pandas as pd
import sqlite3

# Connect to database
conn = sqlite3.connect('retail_sales.db')

# Run query
query = """
SELECT 
    COUNT(*) as total_orders,
    ROUND(SUM(Sales), 2) as total_revenue,
    ROUND(SUM(Profit), 2) as total_profit,
    ROUND(AVG(Sales), 2) as avg_order_value
FROM orders;
"""

result = pd.read_sql(query, conn)
print(" TOTAL SALES OVERVIEW:")
print(result)
```

```python
query = """
SELECT 
    strftime('%Y', "Order Date") as year,
    COUNT(*) as num_orders,
    ROUND(SUM(Sales), 2) as total_sales,
    ROUND(SUM(Profit), 2) as total_profit
FROM orders
GROUP BY year
ORDER BY year;
"""
result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT 
    "Product Name",
    COUNT(*) AS times_sold,
    ROUND(SUM(Sales), 2) AS total_revenue,
    ROUND(SUM(Profit), 2) AS total_profit,
    ROUND(AVG(Sales), 2) AS avg_price
FROM orders
GROUP BY "Product Name"
ORDER BY total_revenue DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """ 
SELECT 
    Category,
    COUNT(*) as num_orders,
    ROUND(SUM(Sales), 2) as total_sales,
    ROUND(SUM(Profit), 2) as total_profit,
    ROUND((SUM(Profit) / SUM(Sales)) * 100, 2) as profit_margin_pct
FROM orders
GROUP BY Category
ORDER BY total_sales DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT 
    "Customer Name",
    COUNT(*) as num_orders,
    ROUND(SUM(Sales), 2) as total_spent,
    ROUND(AVG(Sales), 2) as avg_order_value,
    ROUND(SUM(Profit), 2) as profit_generated
FROM orders
GROUP BY "Customer Name"
ORDER BY total_spent DESC
LIMIT 10;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT 
    "Ship Mode",
    COUNT(*) as num_orders,
    ROUND(SUM(Sales), 2) as total_sales,
    ROUND(AVG(Sales), 2) as avg_order_value
FROM orders
GROUP BY "Ship Mode"
ORDER BY num_orders DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT 
    CASE 
        WHEN Discount = 0 THEN 'No Discount'
        WHEN Discount <= 0.1 THEN 'Small (1-10%)'
        WHEN Discount <= 0.2 THEN 'Medium (11-20%)'
        ELSE 'Large (20%+)'
    END as discount_range,
    COUNT(*) as num_orders,
    ROUND(AVG(Sales), 2) as avg_sales,
    ROUND(AVG(Profit), 2) as avg_profit,
    ROUND((SUM(Profit) / SUM(Sales)) * 100, 2) as profit_margin_pct
FROM orders
GROUP BY discount_range
ORDER BY avg_profit DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT 
    "Sub-Category",
    Category,
    COUNT(*) as num_orders,
    ROUND(SUM(Sales), 2) as total_sales,
    ROUND(SUM(Profit), 2) as total_profit,
    ROUND((SUM(Profit) / SUM(Sales)) * 100, 2) as profit_margin_pct
FROM orders
GROUP BY "Sub-Category", Category
ORDER BY total_sales DESC
LIMIT 20;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
WITH monthly_sales AS (
    SELECT 
        strftime('%Y-%m', "Order Date") as year_month,
        ROUND(SUM(Sales), 2) as total_sales
    FROM orders
    GROUP BY year_month
)
SELECT 
    year_month,
    total_sales,
    LAG(total_sales) OVER (ORDER BY year_month) as previous_month_sales,
    ROUND(total_sales - LAG(total_sales) OVER (ORDER BY year_month), 2) as sales_change,
    ROUND(((total_sales - LAG(total_sales) OVER (ORDER BY year_month)) / 
           LAG(total_sales) OVER (ORDER BY year_month)) * 100, 2) as growth_percentage
FROM monthly_sales
ORDER BY year_month;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
WITH customer_metrics AS (
    SELECT 
        "Customer Name",
        -- Recency: Days since last purchase
        JULIANDAY('2017-12-31') - JULIANDAY(MAX("Order Date")) as days_since_last_order,
        -- Frequency: Number of orders
        COUNT(*) as num_orders,
        -- Monetary: Total spent
        ROUND(SUM(Sales), 2) as total_spent
    FROM orders
    GROUP BY "Customer Name"
),
customer_scores AS (
    SELECT 
        *,
        NTILE(5) OVER (ORDER BY days_since_last_order) as recency_score,
        NTILE(5) OVER (ORDER BY num_orders DESC) as frequency_score,
        NTILE(5) OVER (ORDER BY total_spent DESC) as monetary_score
    FROM customer_metrics
)
SELECT 
    "Customer Name",
    days_since_last_order,
    num_orders,
    total_spent,
    recency_score,
    frequency_score,
    monetary_score,
    (recency_score + frequency_score + monetary_score) as rfm_score,
    CASE 
        WHEN (recency_score + frequency_score + monetary_score) >= 13 THEN 'Champions'
        WHEN (recency_score + frequency_score + monetary_score) >= 10 THEN 'Loyal Customers'
        WHEN (recency_score + frequency_score + monetary_score) >= 7 THEN 'Potential Loyalists'
        WHEN recency_score <= 2 THEN 'At Risk'
        ELSE 'Needs Attention'
    END as customer_segment
FROM customer_scores
ORDER BY rfm_score DESC;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
WITH ranked AS (
    SELECT 
        Category,
        "Sub-Category",
        "Product Name",
        ROUND(SUM(Sales), 2) AS total_sales,
        ROUND(SUM(Profit), 2) AS total_profit,
        ROW_NUMBER() OVER (
            PARTITION BY Category 
            ORDER BY SUM(Profit) DESC
        ) AS profit_rank
    FROM orders
    GROUP BY Category, "Sub-Category", "Product Name"
)
SELECT *
FROM ranked
WHERE profit_rank <= 5
ORDER BY Category, profit_rank;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT 
    "Customer Name",
    MIN("Order Date") as first_purchase,
    MAX("Order Date") as last_purchase,
    COUNT(*) as total_orders,
    ROUND(SUM(Sales), 2) as lifetime_value,
    ROUND(AVG(Sales), 2) as avg_order_value,
    ROUND(SUM(Profit), 2) as total_profit_generated,
    CAST((JULIANDAY(MAX("Order Date")) - JULIANDAY(MIN("Order Date"))) AS INTEGER) as customer_lifespan_days
FROM orders
GROUP BY "Customer Name"
ORDER BY lifetime_value DESC
LIMIT 50;
"""
result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT 
    strftime('%Y', "Order Date") as year,
    COUNT(*) as total_orders,
    ROUND(SUM(Sales), 2) as total_sales,
    ROUND(SUM(Profit), 2) as total_profit,
    LAG(ROUND(SUM(Sales), 2)) OVER (ORDER BY strftime('%Y', "Order Date")) as previous_year_sales,
    ROUND(((SUM(Sales) - LAG(SUM(Sales)) OVER (ORDER BY strftime('%Y', "Order Date"))) / 
           LAG(SUM(Sales)) OVER (ORDER BY strftime('%Y', "Order Date"))) * 100, 2) as yoy_growth_pct
FROM orders
GROUP BY year
ORDER BY year;
"""
result = pd.read_sql_query(query, conn)
print(result)
```
